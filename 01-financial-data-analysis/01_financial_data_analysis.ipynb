# Analyze cryptocoins prices

Download history prices (parameters can be changed like list of coins, timeframe, period)

```python
import ccxt
import pandas as pd
import matplotlib.pyplot as plt
```

```python
exchange = ccxt.binance()
timeframe = "1d"
period = 365
symbols = ["BTC/USDT", "ETH/USDT", "SOL/USDT", "XRP/USDT", "DOGE/USDT", "LTC/USDT"]

data_map = {}
for symbol in symbols:
    data = exchange.fetch_ohlcv(symbol, timeframe=timeframe, limit=period)
    df = pd.DataFrame(
        data,
        columns=["time", "open", "high", "low", "close", "volume"]
    )
    df["time"] = pd.to_datetime(df["time"], unit="ms")
    df = df.set_index("time")
    data_map[symbol] = df
    print(symbol)
    print(df.tail())
```

What happens if we invest 1000 usdt to coins (simple strategy buy and hold)

```python
amount_to_invest_usdt = 1000
for symbol in symbols:
    df = data_map[symbol]
    start_price = df["close"].iloc[0]
    coins_invested = amount_to_invest_usdt / start_price
    df['cost_of_invested'] = df["close"] * coins_invested
fig, ax = plt.subplots(figsize=(10, 7))

for symbol in symbols:
    df = data_map[symbol]
    df["cost_of_invested"].plot(ax=ax, label=symbol)

ax.axhline(amount_to_invest_usdt, color="gray", linestyle="--")
ax.set_title("Value of 1000 USDT invested at the start")
ax.set_ylabel("USDT")
ax.legend()
plt.tight_layout()
plt.show()
```
